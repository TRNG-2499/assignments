## Analysis

Initially the model didn't even label the individual reviews but tallied the results of the categorisation and returned that instead, analysing how the reviews themselves were written. This gave me the idea to use the four labels it returned as my categories (Positive, Negative, Mixed and Neutral) since they were quite effective at encompassing all the reviews, where initially I had planned to use one neutral category to capture both mixed opinions and non-reviews. The second prompt incorporated those categories as explicit instructions as well as the reason column and individual review labelling, though without much formatting since I was curious to see how the model would return the data now that it explicitly had to modify it to some degree. It gave quite a long-winded response and returned a CSV table with all the original columns, and I thought markdown would be more readable and transferrable. The third version was the biggest change, since I incorporated more contraints on output including which columns to include and exclude, and the format of the response. Asking for just the table made the response much simpler. Version 4 introduced role-prompting and few-shot prompting slightly refine the output. This was the most subtle change, only resulting in slightly more detailed reasons given for each, and review 005 was interpreted slightly more negative by version 4 than version 3, which rated it as Mixed.
