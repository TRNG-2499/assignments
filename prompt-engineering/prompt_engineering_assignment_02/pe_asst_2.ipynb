## 3. Analysis

### 1. Which techniques you used, and where in your prompts you used them

I used four core prompt engineering techniques in a structured progression:

* **Role Framing:** Defined the AI's persona as an e-commerce customer feedback analyst to set the professional tone.
* **Defined Categories:** Clearly established three distinct categories (`Positive`, `Negative`, `Mixed`) with explicit definitions to prevent vague or overlapping interpretations.
* **Few-Shot Examples:** Provided clear examples demonstrating tricky edge cases, such as handling sarcasm and balanced critiques.
* **Output Format & Constraints:** Mandated a markdown table structure with specific columns, word count limits for reasons, and an explicit rule to process all 110 reviews without stopping early.

---

### 2. How each technique affected the output (with specific version comparisons)

* **Role & Defined Categories:** 
  * *Version 1* simply summarized the text in paragraph form, making sorting impossible. 
  * *Version 2* introduced explicit categories, forcing the model to classify each item rather than writing a vague overview.

* **Output Format:** 
  * *Version 3* added table formatting requirements. This transformed messy text into structured tabular data ready for spreadsheet analysis, though truncation remained an issue.

* **Few-Shot Examples & Constraints:** 
  * *Version 4* included edge-case examples and explicit completion rules, which enabled the model to accurately catch sarcastic phrasing and process all 110 reviews without cutting off.

---

### 3. How the model handled tricky reviews, and what I changed in response

* **What went wrong:** Initially, the model struggled with sarcastic reviews (e.g., treating "Oh great, it broke immediately" as positive because of the word 'great') and tended to stop halfway through the 110 reviews.
* **How I fixed it:** I added a sarcastic example in the few-shot section (`"Sure, I LOVE cleaning up leaks every day." -> Negative`) and added a strong constraint reminding the model to process every single review through to the end of the dataset.

---

### 4. Which technique made the biggest difference, and why

**Output Format combined with Defined Categories** made the biggest difference.

While role and few-shot examples helped refine accuracy on tricky reviews, forcing a structured markdown table transformed an unstructured LLM chat output into actual usable business data. Without strict formatting rules, the model defaults to conversational paragraphs that require manual parsing.
