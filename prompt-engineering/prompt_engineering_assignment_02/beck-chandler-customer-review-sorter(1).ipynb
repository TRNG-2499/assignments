Prompt 2: You are a marketing manager, going over reviews for your companies listed products. go over each review and determine which reviews are positive, negative, mixed, neutral, and which ones you are unsure about. Each review must receive exactly two labels, and you must return me a csv file containing the review_id, the product, the review itself, and a column containing your label.

note: in an attempt to confuse the model, I've told it to assign 2 labels to each review. Assigning 2 labels to the reviews makes labeling uncertain and mixed reviews easier as you can identify positive and negative parts of the review and label accordingly, while making labeling very positive or very negative reviews harder since there are less uncertenties in those reviews. This was done purely to see if the model can recognize unclear or difficult-to-follow instructions amidst an otherwise clear prompt.

### CSV structure

The file contains four columns: `review_id`, `product`, `review_text`, and `labels`.

Each review receives exactly two labels in the final column:

- Sentiment: positive, negative, mixed, neutral, or unsure.
- Confidence: high_confidence, medium_confidence, or low_confidence.

Example: `negative | high_confidence`

Sarcasm, contradictory feedback, and ambiguous language were considered during classification.

comments: instead of assigning two sentement labels like I was expecting, the model instead created a new column called 'confidence', and used that as the 2nd label. LLM models have a tendancy to work around certain phrasing in order to solve a given problem.
