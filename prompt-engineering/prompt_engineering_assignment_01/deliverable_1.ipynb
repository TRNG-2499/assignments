
### Analysis

For the water bottle, I first integrated role prompting and context to align the model more closely with the specific idea of a product i had in mind, especially considering how vague the original prompt was and, consequently, how vague the first response was. It was interesting how it came up with details for the description while remaining as vague as it could. You might get the sense that it was describing a steel or hard plastic bottle but no mention of material is made besides in comparison to single-use plastic bottles. It also didn't occur to me that the model would simply return a list of features tacked on to the end in an effort to flesh it out. It did this again once I prompted with specific features which aligned it more closely with my idea of the bottle, so in the next pass I integrated constraints and output formatting to structure the output in the way I preferred.
This probably made the biggest difference between consecutive versions, as with parameters like tone, length and structure defined the model had a lot more to work with. In the final pass I added an example of the format I wanted and actully liked this version less, partially because of how closely it adhered to the provided format. I can see how useful that cohesion would be in the case of writing code, but for other kinds of writing, it's a little too derivative compared to the previous output.
