
#### Water Bottle – Version 3 (Constraints & Output Format)
* **Prompt:**
  ```text
  You are an experienced copywriter for an eco-friendly outdoor gear startup.
  
  Write a product description for our reusable water bottle. The target audience consists of environmentally conscious backpackers and hikers. This description will be used on our primary website.
  
  Rules & Constraints:
  - Length: Under 60 words
  - Avoid generic marketing jargon like "ultimate," "perfect companion," or "sleek."

  ```
* **Model Output:**
  ```text
  **Keep the Wild Wild**
  
  This canteen is built from 100% recycled steel to handle drops on mountain rocks. Vacuum insulation keeps your creek water cold all day long. Hydrate sustainably with a simple tool designed to outlast the trail.
  ```

### Part 3: Apply Your Techniques to New Products


## 2. Analysis

### 1. Which techniques you used, and where in your prompts you used them

I used three main techniques, organized in a specific order:

* **Role & Context (Top of the prompt):** Set the AI's role and target audience right away (e.g., *"You are a copywriter for an eco-friendly outdoor gear startup"* or *"a tech copywriter for a minimalist home brand"*).
* **Constraints & Formatting Rules (Middle):** Used a bulleted list to set clear rules like word count limits, banned hype words (like "ultimate" or "sleek"), and required structures (like ending with a *"Best for..."* sentence).
* **Few-Shot Examples (Bottom):** Placed a sample of our brand's "house style" right at the end to show the model the exact tone and format I wanted.

---

### 2. How each technique affected the output (with specific version comparisons)

* **Role & Context:**
  * *Version 1 (Simple Prompt)* produced a generic response: *"Stay hydrated on the go with our premium water bottle!"*
  * *Version 2* added context, which immediately upgraded the vocabulary to outdoor-focused terms like *"100% recycled stainless steel"* and *"sun-drenched miles"*.

* **Constraints & Formatting:**
  * *Version 3* added word limits and banned buzzwords, which forced the AI to strip out fluff. The output became tight and punchy: *"This canteen is built from 100% recycled steel to handle drops on mountain rocks..."*

* **Few-Shot Examples:**
  * *Version 4* included a sample description. This guided the model to match our brand's exact sentence flow, pacing, and ending style without needing long explanations.

---

### 3. Times the model ignored parts of the prompt or hallucinated details, and how I fixed it

* **What went wrong:** When prompts were vague, the model made up features—like claiming a thermostat was made of "brushed-finished steel" or that coffee arrived in "biodegradable bags." It also tended to ignore word count limits when they were buried in paragraphs instead of lists.
* **How I fixed it:** I added a clear **Key Features** list for facts and included a negative constraint telling the AI not to add any features outside that list. I also kept all rules in bulleted format.

---

### 4. Which technique made the biggest difference, and why

**Few-Shot Prompting** made the biggest difference.

Rules and roles act like boundaries—they tell the model what to avoid. Showing the model an actual example acts like a blueprint. Giving it just one sentence of our real house style helped it grab the right tone and cadence instantly, giving the copy a much more natural, human feel.
