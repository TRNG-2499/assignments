## PART 3.5: Reflection

- For both prompts in part three, I used role prompting and constraints and then I used context for the standing desk and output format for the snake plant. I would put the role first, then context/output format, and finally any constraints.
- Both improved prompts changed the length of the output and followed the prompt specifications exactly. The specific differences include how the output becomes less generic and more pointed to a specific audience, the length of the output, and tone.
- The model did not ignore any part of my prompts and I didn't need to edit my prompts.
- I think that the role prompting technique made the biggest difference because the tone and information is affected by the role you give the model.
