I will be using the the prompting techniques of Role Prompting, adding context, and by adding constraints.

### Prompt 1: ###
``- You are an advertising copywriter working with an electronics company to create a product description for a new pair of over-ear noise-canceling headphones. The headphones are marketed toward college students who study in shared spaces and regularly commute. The product features active noise cancellation, a transparency mode, up to 30 hours of battery life, a foldable design, and Bluetooth multipoint connectivity. Write a product description between 180 and 220 words. Include a headline, a short introduction, a list of key features, and a closing slogan. Keep the tone energetic and practical, and do not invent additional technical specifications or claim that the headphones eliminate all noise.``

This prompt assigns the LLM a role, gives it a more specific product and intended audience, and includes constraints for the length, format, and types of claims that can be made.

Keep the description between 160 and 200 words, use a headline followed by a short introduction and four feature bullets, and finish with a slogan. Do not promise a specific reduction in energy bills, claim compatibility with every heating and cooling system, or invent additional product features.

The length and tone constraints led to responses that, while tonally matches what was specified in the prompt, are still very similar to the first couple prompts we ran in steps 1 and 2. The responses follow typically overused styles by AI. These primary style in these responses is the 'Rule of Three' where responses will give examples or lists in three's. Examples in our responses are, 

    ``..uses energy, identify patterns, and make informed decisions..``
    ``..Power through long study sessions, daily commutes, and packed schedules..``
    ``..designed for everyday convenience, durability, and effortless style..``
These examples reflect response patterns that have existed since the broad introduction of LLM's in the early 2020's (rule of three, overuse of em-dashes. The difference in prompting has resulted in different inclusions and formats based on whatever we add to the prompt, but the responses continue to contain a lot of the more subtle LLM-isms that are seen in many other use cases for generative AI.
